```python
from ast import Index
from numpy import random as rng
N_IMAGES = 24
SEED = 2026

selected_parts = []

for condition, condition_data in images.groupby("condition"):
  print(condition)
  selected_condition  = condition_data.sample(n = N_IMAGES, replace=False, random_state=SEED).copy() # selecting 24 images at random
  selected_condition['block'] = np.repeat([1,2,3,4], 6)
  selected_parts.append(selected_condition)

combined_data = pd.concat(selected_parts, ignore_index = True)

# randomizing trial orders again in block

randomized_blocks = []

for block_number in range(1,5):

  block = combined_data[combined_data['block'] == block_number].copy()
  random_order = rng.permutation(len(block))

  block = block.iloc[random_order].reset_index(drop = True)

  block['trial_in_block'] = np.arange(1, 31)

  randomized_blocks.append(block)

selected = pd.concat(randomized_blocks, ignore_index= True)

selected['schedule'] = SEED
selected['trial_number'] = np.arange(1,121)


selected.to_csv(f'file_{SEED}.csv', index = False)


```
